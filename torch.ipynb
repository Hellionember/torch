```python
# Исходные данные (например, x = площадь квартиры, ideal_y = реальная цена)
x = torch.tensor(2.0)
ideal_y = torch.tensor(10.0)

# Наши параметры, которые сеть будет обучать. Задаем им requires_grad=True!
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
```

```python
# 1. Прямой проход (Forward pass) - сеть делает предсказание
predictet_y = w * x + B

# 2. Считаем ошибку (Loss) - насколько мы далеки от идеала (Используем MSE - квадрат ошибки)
loss = (predictet_y - ideal_y)**2

# УСРЕДНЯЕМ ОШИБКУ до одного числа (скаляра) перед backward
loss = loss.mean()

# 3. МАГИЯ PYTORCH: считаем градиенты (Backpropagation)
loss.backward()

# Теперь мы можем посмотреть, куда нам нужно менять w и b
print('Градиент для w:', w.grad)
print('Градиент для b:', b.grad)
```
